# Project - Airline AI Assistant

We'll now bring together what we've learned to make an AI Customer Support assistant for an Airline

```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
# Initialization

load_dotenv(override=True)

openai_api_key = os.getenv('OPENAI_API_KEY')
if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
    
MODEL = "gpt-4.1-mini"
openai = OpenAI()

# As an alternative, if you'd like to use Ollama instead of OpenAI
# Check that Ollama is running for you locally (see week1/day2 exercise) then uncomment these next 2 lines
# MODEL = "llama3.2"
# openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
```

```python
system_message = """
You are a helpful assistant for an Airline called FlightAI.
Give short, courteous answers, no more than 1 sentence.
Always be accurate. If you don't know the answer, say so.
"""
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=MODEL, messages=messages)
    return response.choices[0].message.content

gr.ChatInterface(fn=chat, type="messages").launch()
```

## Tools

Tools are an incredibly powerful feature provided by the frontier LLMs.

With tools, you can write a function, and have the LLM call that function as part of its response.

Sounds almost spooky.. we're giving it the power to run code on our machine?

Well, kinda.

```python
# Let's start by making a useful function

ticket_prices = {"london": "$799", "paris": "$899", "tokyo": "$1400", "berlin": "$499"}

def get_ticket_price(destination_city):
    print(f"Tool called for city {destination_city}")
    price = ticket_prices.get(destination_city.lower(), "Unknown ticket price")
    return f"The price of a ticket to {destination_city} is {price}"
```

```python
get_ticket_price("London")
```

```python
# There's a particular dictionary structure that's required to describe our function:

price_function = {
    "name": "get_ticket_price",
    "description": "Get the price of a return ticket to the destination city.",
    "parameters": {
        "type": "object",
        "properties": {
            "destination_city": {
                "type": "string",
                "description": "The city that the customer wants to travel to",
            },
        },
        "required": ["destination_city"],
        "additionalProperties": False
    }
}
```

```python
# Let's start by making a useful function

weathers = {"london": "Nice and Sunny.", "paris": "Rainy and cosy!", "tokyo": "Snowing and cool!", "berlin": "Sunny and cool!"}

def get_weather(destination_city):
    print(f"Tool called for weather of city {destination_city}")
    weather = weathers.get(destination_city.lower(), "Unknown ticket price")
    return f"The weather in {destination_city} is {weather}"
```

```python
get_weather("London")
```

```python
weather_function = {
    "name": "get_weather",
    "description": "Get the weather of the destination city.",
    "parameters": {
        "type": "object",
        "properties": {
            "destination_city": {
                "type": "string",
                "description": "The city that the customer wants to travel to",
            },
        },
        "required": ["destination_city"],
        "additionalProperties": False
    }
}
```

```python
# And this is included in a list of tools:

tools = [{"type": "function", "function": price_function}, {"type": "function", "function": weather_function}]
```

```python
print(tools[1])
```

## Getting OpenAI to use our Tool

There's some fiddly stuff to allow OpenAI "to call our tool"

What we actually do is give the LLM the opportunity to inform us that it wants us to run the tool.

Here's how the new chat function looks:

```python
# We have to write that function handle_tool_call:

def handle_tool_call(message):
    tool_call = message.tool_calls[0]
    if tool_call.function.name == "get_ticket_price":
        arguments = json.loads(tool_call.function.arguments)
        city = arguments.get('destination_city')
        price_details = get_ticket_price(city)
        response = {
            "role": "tool",
            "content": price_details,
            "tool_call_id": tool_call.id
        }
    elif tool_call.function.name == "get_weather":
        arguments = json.loads(tool_call.function.arguments)
        city = arguments.get('destination_city')
        weather_details = get_weather(city)
        response = {
            "role": "tool",
            "content": weather_details,
            "tool_call_id": tool_call.id
        }
    return response
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=MODEL, messages=messages, tools=tools)

    if response.choices[0].finish_reason=="tool_calls":
        message = response.choices[0].message
        response = handle_tool_call(message)
        messages.append(message)
        messages.append(response)
        response = openai.chat.completions.create(model=MODEL, messages=messages)
    
    return response.choices[0].message.content
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```

## Let's make a couple of improvements

Handling multiple tool calls in 1 response

Handling multiple tool calls 1 after another

```python
def handle_tool_calls(message):
    responses = []
    for tool_call in message.tool_calls:
        if tool_call.function.name == "get_ticket_price":
            arguments = json.loads(tool_call.function.arguments)
            city = arguments.get('destination_city')
            price_details = get_ticket_price(city)
            responses.append({
                "role": "tool",
                "content": price_details,
                "tool_call_id": tool_call.id
            })
        if tool_call.function.name == "get_weather":
            arguments = json.loads(tool_call.function.arguments)
            city = arguments.get('destination_city')
            weather_details = get_weather(city)
            responses.append({
                "role": "tool",
                "content": weather_details,
                "tool_call_id": tool_call.id
            })
    return responses
```

```python
# above code in a simplified manner

def handle_tool_calls(message):
    responses = []
    function_map = {
        "get_ticket_price": get_ticket_price,
        "get_weather": get_weather
    }

    for tool_call in message.tool_calls:
        func = function_map.get(tool_call.function.name)
        if func:
            args = json.loads(tool_call.function.arguments)
            city = args.get("destination_city")
            details = func(city)
            responses.append({
                "role": "tool",
                "content": details,
                "tool_call_id": tool_call.id
            })

    return responses
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=MODEL, messages=messages, tools=tools)

    """while is put instead of if and in the loop tools=tools call is included for sequential tool call
        like condition based ticket price of another city if London ticket proce is less than 1000$.
    """
    while response.choices[0].finish_reason=="tool_calls":
        message = response.choices[0].message
        responses = handle_tool_calls(message)
        messages.append(message)
        messages.extend(responses)
        response = openai.chat.completions.create(model=MODEL, messages=messages, tools=tools)
    
    return response.choices[0].message.content
```

```python
gr.ChatInterface(fn=chat, type="messages").launch()
```

```python
import sqlite3
```

```python
DB = "prices.db"

with sqlite3.connect(DB) as conn:
    cursor = conn.cursor()
    cursor.execute('CREATE TABLE IF NOT EXISTS prices (city TEXT PRIMARY KEY, price REAL)')
    cursor.execute('CREATE TABLE IF NOT EXISTS weathers (city TEXT PRIMARY KEY, weather TEXT)')
    conn.commit()
```

```python
def get_ticket_price(city):
    print(f"DATABASE TOOL CALLED: Getting price for {city}", flush=True)
    with sqlite3.connect(DB) as conn:
        cursor = conn.cursor()
        cursor.execute('SELECT price FROM prices WHERE city = ?', (city.lower(),))
        result = cursor.fetchone()
        return f"Ticket price to {city} is ${result[0]}" if result else "No price data available for this city"
```

```python
get_ticket_price("London")
```

```python
def set_ticket_price(city, price):
    with sqlite3.connect(DB) as conn:
        cursor = conn.cursor()
        cursor.execute('INSERT INTO prices (city, price) VALUES (?, ?) ON CONFLICT(city) DO UPDATE SET price = ?', (city.lower(), price, price))
        conn.commit()
```

```python
ticket_prices = {"london":799, "paris": 899, "tokyo": 1420, "sydney": 2999}
for city, price in ticket_prices.items():
    set_ticket_price(city, price)
```

```python
get_ticket_price("Tokyo")
```

```python
def get_weather(city):
    print(f"DATABASE TOOL CALLED: Getting weather for {city}", flush=True)
    with sqlite3.connect(DB) as conn:
        cursor = conn.cursor()
        cursor.execute('SELECT weather FROM weathers WHERE city = ?', (city.lower(),))
        result = cursor.fetchone()
        return f"Weather in {city} is {result[0]}" if result else "No price data available for this city"
```

```python
get_weather("London")
```

```python
def set_city_weather(city, weather):
    with sqlite3.connect(DB) as conn:
        cursor = conn.cursor()
        cursor.execute('INSERT INTO weathers (city, weather) VALUES (?, ?) ON CONFLICT(city) DO UPDATE SET weather = ?', (city.lower(), weather, weather))
        conn.commit()
```
